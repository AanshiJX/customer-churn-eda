### Tenure and Churn

Customer tenure is compared between churned and retained customers to
determine whether the length of the customer relationship is associated
with churn.

```python
tenure_churn = data.loc[(data['tenure'] < 12) & (data['Churn'] == 'Yes')]['customerID'].count()
tenure_churn
```

```python
total_churn = data.loc[data['tenure'] < 12]['customerID'].count()
total_churn
```

```python
churned_tenure = tenure_churn / total_churn * 100
churned_tenure
```

```python
tenure_churn_12 = data.loc[(data['tenure'] >= 12) & (data['Churn'] == 'Yes')]['customerID'].count()
tenure_churn_12
```

```python
total_churned = data.loc[data['tenure'] >= 12]['customerID'].count()
total_churned
```

```python
tenure_12 = tenure_churn_12 / total_churned * 100
tenure_12
```

Churned customers have substantially lower average tenure than customers
who remained, indicating that newer customers may be more vulnerable to churn.

### Monthly Charges and Churn

Average monthly charges are compared between customers who churned and
customers who remained to identify whether pricing levels are associated
with churn.

```python
monthly_lessthan_50 = data.loc[(data['MonthlyCharges'] < 50) & (data['Churn'] == 'Yes')]['customerID'].count()
monthly_lessthan_50
```

```python
total_monthly_50 = data.loc[data['MonthlyCharges'] < 50]['customerID'].count()
total_monthly_50
```

```python
monthly_churn_50 = monthly_lessthan_50 / total_monthly_50 * 100
monthly_churn_50
```

```python
monthly_lessthan_100 = data.loc[(data['MonthlyCharges'] >= 50) & (data['MonthlyCharges'] < 100) & (data['Churn'] == 'Yes')]['customerID'].count()
monthly_lessthan_100
```

```python
total_monthly_100 = data.loc[(data['MonthlyCharges'] >= 50) & (data['MonthlyCharges'] < 100)]['customerID'].count()
total_monthly_100
```

```python
monthly_churn_100 = monthly_lessthan_100 / total_monthly_100 * 100
monthly_churn_100
```

```python
monthly_greaterthan_100 = data.loc[(data['MonthlyCharges'] >=  100) & (data['Churn'] == 'Yes')]['customerID'].count()
monthly_greaterthan_100
```

```python
total_monthly_greater_100 = data.loc[data['MonthlyCharges'] >= 100]['customerID'].count()
total_monthly_greater_100
```

```python
monthly_churn_morethan_100 = monthly_greaterthan_100 / total_monthly_greater_100 * 100
monthly_churn_morethan_100
```

Churned customers have a higher average monthly charge than customers
who remained, suggesting an association between monthly charges and churn.

## 7. Conclusion

In this project, I explored customer churn using Pandas and NumPy.

I looked at different factors such as tenure, monthly charges, contract
type, payment method, internet service, and customer details to understand
which groups had higher churn rates.

Some clear patterns were found. Customers with less than 12 months of
tenure had a much higher churn rate of 48.29% compared to 17.49% for
customers with longer tenure. Month-to-month customers also had a much
higher churn rate than customers with one-year or two-year contracts.

Electronic check and fiber optic customers also showed higher churn rates
compared to other groups. Senior citizens also had a higher churn rate
than non-senior customers.

Overall, this analysis helped me understand the main patterns in customer
churn and gave me a better understanding of how Pandas and NumPy can be
used to analyze real-world data.
